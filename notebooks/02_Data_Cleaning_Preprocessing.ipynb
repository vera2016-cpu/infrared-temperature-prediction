# Data Cleaning and Preprocessing

## Objective

The objective of this notebook is to prepare the raw infrared thermography dataset for machine learning.

The preprocessing workflow includes:

- Loading the raw dataset
- Inspecting the dataset
- Handling missing values
- Removing duplicates (if any)
- Correcting data types
- Separating features and targets
- Identifying categorical and numerical features
- Preparing the dataset for feature engineering and modeling

```python
# ============================================
# Import Libraries
# ============================================

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries imported successfully.")
```

```python
# Load the Raw Dataset
df = pd.read_csv("../data/raw.csv")

df.head()
```

```python
df_clean = df.copy()
```

```python
df_clean.info()
```

```python
# Missing Value Analysis
missing = df_clean.isnull().sum()

missing[missing > 0]
```

```python
# Percentage of missing values
missing_percentage = (df_clean.isnull().sum() / len(df_clean)) * 100

missing_percentage[missing_percentage > 0]
```

```python
# Calculate the median
median_distance = df_clean["Distance"].median()

# Replace missing values
df_clean["Distance"] = df_clean["Distance"].fillna(median_distance)
```

```python
print(df_clean["Distance"].isnull().sum())
```

```python
# Check for Duplicate Rows
duplicates = df_clean.duplicated().sum()

print(f"Duplicate rows: {duplicates}")
```

```python
# Data type
df_clean.dtypes
```

```python
# Separate Features and Target
target = "aveOralF"

X = df_clean.drop(columns=["aveOralF", "aveOralM"])

y = df_clean[target]
```

```python
# Shapes of features and target
print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# Identify Feature Types
categorical_features = X.select_dtypes(include="object").columns.tolist()

numerical_features = X.select_dtypes(include=["int64", "float64"]).columns.tolist()
```

```python
print("Categorical Features")
print(categorical_features)

print("\nNumerical Features")
print(numerical_features)
```

```python
# Save the Clean Dataset
cleaned_df = pd.concat([X, y], axis=1)

cleaned_df.to_csv("../data/cleaned.csv", index=False)
```

## Data Cleaning Summary

The following preprocessing steps were completed:

- Loaded the raw dataset
- Created a working copy
- Identified missing values
- Imputed missing values in the `Distance` feature using the median
- Checked for duplicate records
- Verified data types
- Selected `aveOralF` as the prediction target
- Removed `aveOralM` from the feature set to prevent data leakage
- Identified categorical and numerical features
- Saved the cleaned dataset for the next stage of the project
